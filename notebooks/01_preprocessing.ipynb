```python
import pandas as pd
import numpy as np

# Load with correct encoding - UTF-8-SIG for Urdu
df = pd.read_csv("../data/urdu_news.csv", 
                 encoding='utf-8-sig',
                 encoding_errors='replace')

# Fix column names
df.columns = ['Index', 'Headline', 'News Text', 'Category', 'Date', 'URL', 'Source', 'News length']

print("Dataset loaded successfully!")
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())
print("\nSample Headline:")
print(df['Headline'][0])
```

```python
# Drop null rows
df = df.dropna()
df = df.reset_index(drop=True)

print("After cleaning:", len(df), "articles")
print("Nulls remaining:", df.isnull().sum().sum())
print("\nCategory distribution:")
print(df['Category'].value_counts())
```

```python
# Combine Headline + News Text (exactly like ULTRA does)
df['combined_text'] = df['Headline'] + ' ' + df['News Text']

print("Sample combined text:")
print(df['combined_text'][0][:300])
```

```python
# Save the clean dataset
df.to_csv("../data/clean_articles.csv", 
          index=False, 
          encoding='utf-8-sig')

print("✅ Clean file saved!")
print("Location: data/clean_articles.csv")
print("Total articles saved:", len(df))
```
